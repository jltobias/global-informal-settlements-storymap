# Global country index

The index is comprehensive over ISO countries even when settlement geometry is missing. Missing geometry is explicit rather than silently inferred.

```python
from pathlib import Path
import csv, json
ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    if (ROOT.parent / 'data').exists(): ROOT = ROOT.parent
    elif (ROOT / 'content' / 'data').exists(): ROOT = ROOT / 'content'
print('content root:', ROOT.resolve())
```

```python
rows=list(csv.DictReader((ROOT / "data/country_index.csv").open()))
for r in rows[:20]:
    print(r["iso3"], r["country_name"], "features=", r["settlement_feature_count"], "WDI=", r["wdi_slum_share_percent"] or "n/a")
```

Open `global/global_index.jGIS` for the JupyterGIS world entry point. Country-level WDI values are contextual indicators and are intentionally not used to fabricate settlement locations.
