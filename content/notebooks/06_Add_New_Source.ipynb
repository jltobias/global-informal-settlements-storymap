# Add a new source

Use this workflow for every new country/city dataset. The goal is to make provenance auditable before geometry reaches the public storymap.

```python
from pathlib import Path
import csv, json
ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    if (ROOT.parent / 'data').exists(): ROOT = ROOT.parent
    elif (ROOT / 'content' / 'data').exists(): ROOT = ROOT / 'content'
print('content root:', ROOT.resolve())
```

1. Add a record to `sources/source_registry.json`.
2. Place or fetch source GeoJSON into `data/raw/`.
3. Inspect its fields and definition.
4. Create a mapping JSON under `data/mappings/`.
5. Normalize with `scripts/normalize_geojson.py`.
6. Merge normalized features only after checking IDs and duplicate boundaries.
7. Rebuild the country index and `.jGIS` storymaps.
8. Run `scripts/validate.py` and perform the human review in `docs/ETHICS.md`.

For ArcGIS FeatureServer layers, `scripts/fetch_arcgis_geojson.py` can download a layer as GeoJSON when the endpoint supports GeoJSON output.
