# Quality assurance

Publication quality requires both structural validation and substantive review. The automated checks cannot determine whether a source's use of “informal settlement” is socially or methodologically appropriate.

```python
from pathlib import Path
import csv, json
ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    if (ROOT.parent / 'data').exists(): ROOT = ROOT.parent
    elif (ROOT / 'content' / 'data').exists(): ROOT = ROOT / 'content'
print('content root:', ROOT.resolve())
```

```python
fc=json.loads((ROOT / "data/settlements.geojson").read_text())
required={"settlement_id","country_iso3","settlement_name","source_id","source_url","license","geometry_status","representative_lon","representative_lat","bbox_wgs84"}
problems=[]
for i,f in enumerate(fc["features"]):
    p=f.get("properties",{})
    missing=required-set(p)
    if missing: problems.append((i,"missing",sorted(missing)))
    if p.get("geometry_status")=="synthetic" and "synthetic" not in (p.get("settlement_name","")+p.get("notes","")).lower(): problems.append((i,"synthetic labeling",None))
print("problems:", problems or "none")
```

## Human review checklist

- Is the geometry explicitly mapped as an informal settlement/deprived area by the source?
- Is the date appropriate for the story?
- Are uncertainty and omissions documented?
- Is precise publication ethically appropriate?
- Does the license permit redistribution?
- Are community or local-government validations available?
