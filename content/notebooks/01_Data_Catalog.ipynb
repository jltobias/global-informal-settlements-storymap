# Data catalog and coverage

Inspect registered sources, country context, and settlement geometry coverage before adding or publishing data.

```python
from pathlib import Path
import csv, json
ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    if (ROOT.parent / 'data').exists(): ROOT = ROOT.parent
    elif (ROOT / 'content' / 'data').exists(): ROOT = ROOT / 'content'
print('content root:', ROOT.resolve())
```

```python
registry=json.loads((ROOT.parent / "sources/source_registry.json").read_text()) if (ROOT.parent / "sources/source_registry.json").exists() else None
if registry is None:
    print("In JupyterLite, open sources/source_registry.json from the repository ZIP/GitHub checkout for the full source registry.")
else:
    for s in registry["sources"]:
        print(f"{s['id']}: {s['scope']} | geometry={s['geometry']} | {s['license']}")
```

```python
rows=list(csv.DictReader((ROOT / "data/country_index.csv").open()))
with_geom=[r for r in rows if int(r["settlement_feature_count"] or 0)>0]
with_wdi=[r for r in rows if r["wdi_slum_share_percent"] not in ("",None)]
print("countries in index:", len(rows))
print("countries with settlement geometry:", len(with_geom))
print("countries with fetched WDI context:", len(with_wdi))
```
