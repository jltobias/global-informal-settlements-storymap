# Normalize settlement geometry

The canonical feature schema adds provenance, representative latitude/longitude, and WGS84 bounding boxes without changing the original source geometry. The repository script supports any GeoJSON once you supply a property mapping.

````python
from pathlib import Path
import csv, json
ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    if (ROOT.parent / 'data').exists(): ROOT = ROOT.parent
    elif (ROOT / 'content' / 'data').exists(): ROOT = ROOT / 'content'
print('content root:', ROOT.resolve())
````

````python
fc=json.loads((ROOT / "data/settlements.geojson").read_text())
for f in fc["features"][:5]:
    p=f["properties"]
    print(p["settlement_name"], p["representative_lat"], p["representative_lon"], p["bbox_wgs84"], p["geometry_status"])
````

For a new source in a local checkout, run:

```bash
python scripts/normalize_geojson.py data/raw/source.geojson --mapping data/mappings/source.mapping.json --output content/data/settlements.geojson
```

If multiple normalized sources are used, merge their feature arrays only after checking ID uniqueness, definitions, and licenses.
