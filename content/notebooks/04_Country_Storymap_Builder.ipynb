# Country storymap builder

The generator creates one folder per country with three GeoJSON layers—boundary, bounding box, representative point—and a `.jGIS` guided story with one segment per settlement.

````python
from pathlib import Path
import csv, json
ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    if (ROOT.parent / 'data').exists(): ROOT = ROOT.parent
    elif (ROOT / 'content' / 'data').exists(): ROOT = ROOT / 'content'
print('content root:', ROOT.resolve())
````

````python
from collections import Counter
fc=json.loads((ROOT / "data/settlements.geojson").read_text())
counts=Counter(f["properties"]["country_iso3"] for f in fc["features"])
print(counts)
for iso in counts:
    print("storymap:", ROOT / "countries" / iso / f"{iso}.jGIS")
````
